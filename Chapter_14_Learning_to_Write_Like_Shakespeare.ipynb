## 2. Masalah pada Vanilla RNN

Vanilla RNN memiliki masalah ketika sequence sangat panjang:

1. Vanishing Gradient
   → gradient menjadi sangat kecil.
   → model sulit mempelajari hubungan jangka panjang.

2. Exploding Gradient
   → gradient menjadi sangat besar.
   → training dapat menjadi tidak stabil.

Penyebabnya berkaitan dengan perkalian matrix dan nonlinearity
yang dilakukan berulang kali pada setiap timestep.

```python
import numpy as np

sigmoid = lambda x: 1 / (1 + np.exp(-x))
weights = np.array([[1, 4],
                    [4, 1]])

activation = sigmoid(np.array([1, 0.01]))

for _ in range(10):
    activation = sigmoid(activation.dot(weights))
    print(activation)
```

## 3. Long Short-Term Memory (LSTM)

LSTM dibuat untuk mengatasi masalah vanishing/exploding gradient
pada RNN.

Ide utamanya:

LSTM memiliki **cell state (c)** yang berfungsi seperti jalur penyimpanan
informasi jangka panjang.

Informasi dapat:
- disimpan
- dilupakan
- dikeluarkan

Pengaturan tersebut dilakukan menggunakan gates.

## 4. LSTM Gates

LSTM memiliki 3 gates utama:

### Forget Gate (f)
Menentukan informasi lama yang akan dipertahankan atau dilupakan.

### Input Gate (i)
Menentukan informasi baru yang akan dimasukkan.

### Output Gate (o)
Menentukan informasi dari cell state yang digunakan sebagai output.

### Update (u)
Informasi baru yang akan digunakan untuk memperbarui cell state.

## 5. Proses LSTM

Cell state diperbarui dengan:

c = (f × prev_c) + (i × u)

Hidden state:

h = o × tanh(c)

Keterangan:
- c       = cell state baru
- prev_c  = cell state sebelumnya
- f       = forget gate
- i       = input gate
- u       = informasi baru
- o       = output gate
- h       = hidden state

```python
import numpy as np

# Nilai gate
f = 0.8   # forget
i = 0.3   # input
o = 0.7   # output
u = 0.5   # update

prev_c = 1.0

# Cell state baru
c = (f * prev_c) + (i * u)

# Hidden state
h = o * np.tanh(c)

print("Cell state :", c)
print("Hidden state:", h)
```

## 6. Kesimpulan

Vanilla RNN sulit menyimpan informasi jangka panjang karena
mengalami vanishing/exploding gradients.

LSTM mengatasinya dengan menggunakan cell state dan gates.

Konsep penting:

RNN
→ hidden state

LSTM
→ hidden state + cell state
→ forget gate
→ input gate
→ output gate
→ update
