## Artificial Intelligence, Machine Learning, dan Deep Learning

Artificial Intelligence (AI) merupakan bidang yang lebih luas yang
berhubungan dengan kemampuan mesin untuk melakukan tugas yang membutuhkan
kecerdasan.

Machine Learning merupakan bagian dari AI yang berfokus pada kemampuan
mesin untuk belajar dari data dan melakukan suatu tugas tanpa harus
diprogram secara eksplisit untuk setiap kemungkinan kondisi.

Deep Learning merupakan salah satu bagian dari Machine Learning yang
terutama menggunakan artificial neural network.

## Apa Itu Machine Learning?

Machine Learning adalah bidang yang memungkinkan komputer belajar
melakukan suatu tugas tanpa harus diprogram secara eksplisit untuk
setiap kondisi.

Daripada programmer menentukan semua aturan secara manual, sistem
Machine Learning mempelajari pola dari data.

Secara sederhana:

Data → Machine Learning → Pola → Prediksi

Contohnya:

Data:
- ukuran rumah
- jumlah kamar
- lokasi

Machine Learning:
- mempelajari hubungan antara data tersebut dengan harga rumah

Output:
- prediksi harga rumah

## Apa Itu Deep Learning?

Deep Learning merupakan subset dari Machine Learning.

Deep Learning terutama menggunakan artificial neural network untuk
mempelajari pola dari data.

Deep Learning dapat digunakan untuk berbagai jenis data dan permasalahan,
misalnya:

- Gambar → Computer Vision
- Teks → Natural Language Processing
- Suara → Speech Recognition

Jadi, Deep Learning bukan sesuatu yang terpisah dari Machine Learning,
melainkan salah satu pendekatan yang terdapat di dalam Machine Learning.

## Supervised Learning

Supervised Learning adalah jenis Machine Learning yang mempelajari
hubungan antara input dan output.

Pada supervised learning, kita memiliki:

- Input
- Output/target

Model mencoba mempelajari hubungan antara keduanya.

Contoh:

Input:
- luas rumah
- jumlah kamar
- lokasi

Target:
- harga rumah

Model belajar dari pasangan input dan target tersebut sehingga nantinya
dapat memprediksi output untuk data baru.

Secara sederhana:

Input → Model → Prediksi

Kemudian prediksi dibandingkan dengan target yang sebenarnya.

## Contoh Supervised Learning

Misalnya kita ingin membuat model untuk memprediksi apakah sebuah
email merupakan spam atau bukan.

Dataset:

| Isi Email | Label |
|-----------|-------|
| "Anda memenangkan hadiah!" | Spam |
| "Meeting jam 10 pagi" | Bukan Spam |
| "Dapatkan hadiah gratis" | Spam |
| "Laporan tugas sudah dikirim" | Bukan Spam |

Model mempelajari hubungan antara isi email dengan label.

Setelah belajar, model dapat menerima email baru dan memberikan prediksi:

Email baru → Model → Spam / Bukan Spam

## Unsupervised Learning

Unsupervised Learning digunakan ketika kita tidak memiliki target atau
jawaban yang sudah ditentukan.

Tujuan utamanya adalah menemukan pola atau kelompok yang terdapat
di dalam data.

Salah satu contoh paling sederhana adalah clustering.

Contoh:

Kita memiliki data pelanggan tanpa label.

Model dapat menemukan bahwa pelanggan tersebut dapat dibagi menjadi
beberapa kelompok berdasarkan kemiripan karakteristiknya.

Data → Model → Kelompok/Pola

## Contoh Clustering

Misalnya terdapat data mengenai beberapa jenis makanan:

- puppies
- kittens
- pizza
- hot dog
- burger

Algoritma unsupervised dapat menemukan pola dan mengelompokkan data
berdasarkan kemiripan tertentu.

Hasil clustering tidak harus memiliki nama yang diberikan manusia.

Misalnya:

Cluster 1 → makanan yang dianggap lezat
Cluster 2 → hewan yang dianggap lucu

Hal pentingnya adalah algoritma menemukan struktur atau kelompok
dari data tanpa diberikan label terlebih dahulu.

## Supervised vs Unsupervised Learning

| Supervised Learning | Unsupervised Learning |
|---|---|
| Memiliki target/label | Tidak memiliki target yang diketahui |
| Belajar dari input dan output | Mencari pola dari data |
| Prediksi dibandingkan dengan target | Mencari struktur/kelompok |
| Contoh: klasifikasi spam | Contoh: clustering |
| Tujuan: memprediksi output | Tujuan: menemukan pola |

### Cara mudah mengingat

Supervised:

Data + Jawaban → Belajar → Prediksi

Unsupervised:

Data tanpa jawaban → Belajar → Menemukan pola

## Parametric vs Nonparametric Learning

Selain supervised dan unsupervised, Machine Learning juga dapat
dibedakan menjadi:

1. Parametric
2. Nonparametric

Perbedaannya berkaitan dengan bagaimana pembelajaran disimpan dan
berapa banyak parameter yang digunakan oleh model.

## Parametric Learning

Model parametric memiliki jumlah parameter yang tetap.

Sederhananya, kita dapat membayangkan model mempunyai sejumlah
"knob" atau pengaturan yang dapat diubah selama proses pembelajaran.

Contoh sederhana:

Model memiliki:

- parameter 1
- parameter 2
- parameter 3

Jumlah parameter tersebut sudah ditentukan.

Proses belajar dilakukan dengan mengubah nilai parameter tersebut
hingga model menghasilkan prediksi yang lebih baik.

## Nonparametric Learning

Nonparametric learning memiliki jumlah parameter yang bergantung
pada data.

Artinya, jumlah parameter tidak ditentukan sepenuhnya sejak awal.

Jika data bertambah atau struktur data berubah, jumlah parameter
yang digunakan dapat ikut berubah.

Cara mudah mengingat:

Parametric:
Jumlah parameter tetap.

Nonparametric:
Jumlah parameter mengikuti data.

## Perbandingan Parametric dan Nonparametric

| Parametric | Nonparametric |
|---|---|
| Jumlah parameter tetap | Jumlah parameter bergantung pada data |
| Menggunakan sejumlah parameter yang sudah ditentukan | Parameter dapat bertambah berdasarkan data |
| Dapat dibayangkan seperti mengatur sejumlah "knob" | Dapat dibayangkan seperti melakukan counting |
| Deep Learning termasuk model parametric | Tidak menjadi fokus utama buku |

### Catatan

Pembagian parametric dan nonparametric tidak selalu benar-benar
hitam dan putih. Buku juga menjelaskan bahwa terdapat area abu-abu
di antara keduanya.

## Empat Kombinasi Jenis Machine Learning

Supervised/Unsupervised dan Parametric/Nonparametric merupakan
dua pembagian yang berbeda.

Karena itu, keduanya dapat dikombinasikan menjadi empat jenis:

1. Supervised Parametric
2. Unsupervised Parametric
3. Supervised Nonparametric
4. Unsupervised Nonparametric

Secara sederhana:

                    Parametric
                   /           \
          Supervised         Unsupervised

                  Nonparametric
                   /           \
          Supervised         Unsupervised

## Supervised Parametric Learning

Supervised parametric learning menggunakan:

- Input data
- Target/jawaban
- Parameter dengan jumlah tetap

Model membuat prediksi berdasarkan nilai parameter.

Setelah prediksi dibuat, prediksi dibandingkan dengan kondisi sebenarnya.

Jika prediksi salah, parameter disesuaikan agar prediksi berikutnya
menjadi lebih baik.

Alur sederhananya:

Input
  ->
Model + Parameter
  ->
Prediksi
  ->
Bandingkan dengan Target
  ->
Perbarui Parameter

## Predict → Compare → Learn

Salah satu konsep utama dalam chapter ini adalah:

**Predict → Compare → Learn**

### 1. Predict

Model menerima data dan menghasilkan prediksi.

### 2. Compare

Prediksi dibandingkan dengan pola atau hasil yang sebenarnya.

### 3. Learn

Model menggunakan informasi dari perbandingan tersebut untuk
mengubah parameter sehingga dapat menghasilkan prediksi yang lebih baik.

Alurnya:

Data
 ->
Predict
 ->
Prediksi
 ->
Compare
 ->
Kesalahan
 ->
Learn
 ->
Parameter diperbarui
 ->
Prediksi berikutnya menjadi lebih baik

## Contoh Sederhana Predict → Compare → Learn

Misalnya sebuah model digunakan untuk memprediksi apakah suatu tim
akan menang.

### Predict

Model memprediksi:

Prediksi = 80% kemungkinan menang

### Compare

Hasil sebenarnya:

Tim tersebut kalah.

Maka prediksi model tidak sesuai dengan kenyataan.

### Learn

Model menggunakan informasi tersebut untuk mengubah parameter yang
digunakan sehingga pada kondisi yang sama atau mirip, prediksi berikutnya
dapat menjadi lebih akurat.

Jadi:

Prediksi → Bandingkan → Perbaiki → Prediksi lagi

## Unsupervised Parametric Learning

Unsupervised parametric learning menggabungkan dua konsep:

- Unsupervised → mencari kelompok/pola tanpa target yang diketahui.
- Parametric → menggunakan jumlah parameter yang ditentukan.

Model dapat memiliki beberapa parameter untuk setiap kelompok.

Parameter tersebut digunakan untuk menentukan seberapa besar kemungkinan
suatu data termasuk ke dalam kelompok tertentu.

Contohnya jika terdapat tiga kelompok:

Group 1
Group 2
Group 3

Model dapat menghasilkan nilai yang menunjukkan seberapa kuat sebuah
data berhubungan dengan masing-masing kelompok.

## Nonparametric Learning

Pada nonparametric learning, jumlah parameter bergantung pada data.

Salah satu cara sederhana untuk membayangkannya adalah menggunakan
metode counting.

Misalnya sebuah model ingin mempelajari hubungan antara warna lampu
lalu lintas dengan tindakan kendaraan.

Model dapat menghitung:

- Berapa kali lampu tertentu muncul.
- Berapa kali kendaraan berhenti.
- Berapa kali kendaraan berjalan.

Jika terdapat lebih banyak jenis lampu atau kategori yang perlu dihitung,
jumlah parameter juga dapat bertambah.

Jadi:

Jumlah data/struktur yang dipelajari
            ->
Jumlah parameter

```python
# Simulasi sederhana konsep Predict → Compare → Learn

prediksi = 0.8
hasil_sebenarnya = 0

error = prediksi - hasil_sebenarnya

print("Prediksi          :", prediksi)
print("Hasil sebenarnya  :", hasil_sebenarnya)
print("Error             :", error)
```
