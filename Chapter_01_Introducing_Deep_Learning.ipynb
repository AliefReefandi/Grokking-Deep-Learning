## Apa Itu Deep Learning?

Deep Learning merupakan salah satu pendekatan dalam Machine Learning yang
menggunakan neural network sebagai bagian utama dari proses pembelajaran.

Dalam buku *Grokking Deep Learning*, Deep Learning diperkenalkan sebagai
teknologi yang berusaha melakukan otomatisasi kecerdasan secara bertahap.

Deep Learning telah digunakan dalam berbagai bidang, seperti:

- Computer Vision
- Speech Recognition
- Machine Translation
- Natural Language Processing
- dan berbagai tugas lain yang membutuhkan pembelajaran dari data.

Salah satu karakteristik penting Deep Learning adalah penggunaan neural
network yang terinspirasi secara longgar dari cara kerja otak manusia.

## Mengapa Belajar Deep Learning?

Ada beberapa alasan mengapa Deep Learning menarik untuk dipelajari.

### 1. Mengotomatisasi kecerdasan

Deep Learning dapat digunakan untuk membuat sistem yang belajar dari data
sehingga sebagian pekerjaan yang sebelumnya membutuhkan kemampuan manusia
dapat dilakukan secara otomatis.

### 2. Banyak digunakan dalam berbagai bidang

Deep Learning telah memberikan perkembangan yang besar pada berbagai bidang
seperti pengolahan gambar, suara, bahasa, dan berbagai permasalahan lainnya.

### 3. Memiliki potensi untuk mengotomatisasi pekerjaan

Kemampuan Deep Learning dalam mempelajari pola dari data membuat teknologi
ini memiliki potensi untuk membantu atau mengotomatisasi berbagai pekerjaan
yang membutuhkan kemampuan analisis.

### 4. Menarik dan kreatif

Deep Learning tidak hanya digunakan untuk pekerjaan teknis. Teknologi ini
juga dapat digunakan untuk mempelajari pola dalam karya seni, bahasa, dan
berbagai bentuk kreativitas manusia.

## Deep Learning sebagai Otomatisasi Kecerdasan

Salah satu gagasan utama pada chapter ini adalah bahwa Deep Learning dapat
dilihat sebagai alat untuk melakukan otomatisasi kecerdasan secara bertahap.

Manusia telah lama membuat berbagai alat untuk membantu memahami dan
mengendalikan lingkungan di sekitarnya. Deep Learning dapat dipandang sebagai
salah satu perkembangan dari usaha tersebut.

Berbeda dengan alat mekanis biasa, Deep Learning berfokus pada kemampuan
sistem untuk mempelajari pola dan menghasilkan keputusan atau prediksi
berdasarkan data.

Dengan demikian, Deep Learning tidak hanya berkaitan dengan membuat program
yang mengikuti aturan yang ditentukan secara manual, tetapi juga membuat
sistem yang dapat belajar dari data.

## Dampak Deep Learning

Perkembangan Deep Learning memiliki dampak yang cukup besar terhadap
teknologi dan industri.

Beberapa contoh bidang yang mengalami perkembangan melalui Deep Learning
antara lain:

- Computer Vision
- Speech Recognition
- Machine Translation
- Analisis bahasa
- Sistem prediksi
- Otomatisasi berbagai pekerjaan

Teknologi ini juga memiliki potensi untuk mengubah cara manusia melakukan
berbagai pekerjaan.

Namun, perkembangan tersebut juga membawa tantangan, terutama terkait
perubahan kebutuhan tenaga kerja dan kemampuan manusia dalam beradaptasi
dengan teknologi baru.

## Deep Learning dan Kreativitas

Menurut buku, salah satu hal menarik dari Deep Learning adalah kemampuannya
untuk digunakan pada permasalahan yang berhubungan dengan kreativitas.

Deep Learning dapat digunakan untuk mempelajari pola dari data yang
kompleks. Contohnya adalah mempelajari karya seni dan menghasilkan karya
baru berdasarkan pola yang telah dipelajari.

Hal ini menunjukkan bahwa Deep Learning tidak hanya berkaitan dengan
