# Chapter 5 — Learning Multiple Weights at a Time: Generalizing Gradient Descent

## 2. Gradient Descent dengan Multiple Inputs

Pada chapter sebelumnya, gradient descent digunakan untuk memperbarui satu weight.

Sekarang sebuah neural network dapat menerima beberapa input sekaligus.

Contohnya:

- jumlah toes pemain
- persentase pertandingan yang dimenangkan
- jumlah fans

Setiap input memiliki weight masing-masing.

Prediction dihitung dengan menjumlahkan hasil perkalian setiap input
dengan weight-nya.

Rumus sederhananya:

prediction = input₁ × weight₁
           + input₂ × weight₂
           + input₃ × weight₃

```python
def w_sum(a, b):
    assert(len(a) == len(b))

    output = 0

    for i in range(len(a)):
        output += a[i] * b[i]

    return output


def neural_network(input, weights):
    pred = w_sum(input, weights)
    return pred


weights = [0.1, 0.2, -0.1]

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

win_or_lose_binary = [1, 1, 0, 1]

input = [toes[0], wlrec[0], nfans[0]]
true = win_or_lose_binary[0]

pred = neural_network(input, weights)

error = (pred - true) ** 2
delta = pred - true

print("Prediction:", pred)
print("Error:", error)
print("Delta:", delta)
```

## 3. Menghitung Weight Delta

Karena terdapat beberapa input dan weight, maka setiap weight harus memiliki
weight_delta masing-masing.

Konsepnya masih sama seperti chapter sebelumnya:

weight_delta = input × delta

Namun sekarang dilakukan untuk setiap input.

Contohnya:

weight_delta₁ = input₁ × delta
weight_delta₂ = input₂ × delta
weight_delta₃ = input₃ × delta

Setiap weight memiliki weight_delta berbeda karena nilai input-nya berbeda,
meskipun menggunakan delta yang sama.

```python
def ele_mul(number, vector):
    output = [0, 0, 0]

    assert(len(output) == len(vector))

    for i in range(len(vector)):
        output[i] = number * vector[i]

    return output


input = [toes[0], wlrec[0], nfans[0]]

pred = neural_network(input, weights)

error = (pred - true) ** 2
delta = pred - true

weight_deltas = ele_mul(delta, input)

print("Prediction:", pred)
print("Error:", error)
print("Delta:", delta)
print("Weight Deltas:", weight_deltas)
```

## 4. Meng-update Multiple Weights

Setelah mendapatkan weight_delta, setiap weight diperbarui menggunakan alpha.

Rumusnya:

weight = weight - (alpha × weight_delta)

Proses ini dilakukan pada semua weight.

Alpha digunakan untuk mengontrol seberapa besar perubahan weight
pada setiap proses learning.

```python
alpha = 0.01

for i in range(len(weights)):
    weights[i] -= alpha * weight_deltas[i]

print("Updated weights:", weights)
```

## 5. Beberapa Iterasi Learning

Neural network dapat melakukan proses prediction, comparison, dan learning
berulang kali.

Pada setiap iterasi:

1. Network membuat prediction.
2. Prediction dibandingkan dengan nilai sebenarnya.
3. Error dan delta dihitung.
4. Weight_delta dihitung.
5. Weight diperbarui.
6. Proses diulang.

```python
def neural_network(input, weights):
    out = 0

    for i in range(len(input)):
        out += input[i] * weights[i]

    return out


def ele_mul(scalar, vector):
    out = [0, 0, 0]

    for i in range(len(out)):
        out[i] = vector[i] * scalar

    return out


toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

win_or_lose_binary = [1, 1, 0, 1]

true = win_or_lose_binary[0]
alpha = 0.01

weights = [0.1, 0.2, -0.1]

input = [toes[0], wlrec[0], nfans[0]]

for iteration in range(3):

    pred = neural_network(input, weights)

    error = (pred - true) ** 2

    delta = pred - true

    weight_deltas = ele_mul(delta, input)

    print("Iteration:", iteration + 1)
    print("Prediction:", pred)
    print("Error:", error)
    print("Delta:", delta)
    print("Weights:", weights)
    print("Weight Deltas:", weight_deltas)
    print()

    for i in range(len(weights)):
        weights[i] -= alpha * weight_deltas[i]
```

## 6. Freezing One Weight

Pada bagian ini, salah satu weight dibuat tetap dan tidak diperbarui.

Tujuannya adalah untuk melihat bagaimana weight lainnya dapat tetap
belajar untuk mengurangi error.

Weight pertama dibuat "frozen" dengan mengubah weight_delta-nya menjadi 0.

Hal ini menunjukkan bahwa setiap weight tidak bekerja secara terpisah.
Semua weight berkontribusi terhadap error yang sama.

```python
weights = [0.1, 0.2, -0.1]

alpha = 0.3

input = [toes[0], wlrec[0], nfans[0]]
true = win_or_lose_binary[0]

for iteration in range(3):

    pred = neural_network(input, weights)

    error = (pred - true) ** 2

    delta = pred - true

    weight_deltas = ele_mul(delta, input)

    # Membekukan weight pertama
    weight_deltas[0] = 0

    print("Iteration:", iteration + 1)
    print("Prediction:", pred)
    print("Error:", error)
    print("Weights:", weights)

    for i in range(len(weights)):
        weights[i] -= alpha * weight_deltas[i]
```

## 7. Gradient Descent dengan Multiple Outputs

Neural network juga dapat menghasilkan beberapa prediction dari satu input.

Setiap output memiliki weight sendiri.

Pada proses learning, masing-masing output mempunyai delta sendiri.

Namun semua weight menggunakan input yang sama.

Secara sederhana:

weight_delta = input × delta

Proses tersebut dilakukan untuk setiap output.

```python
weights = [0.3, 0.2, 0.9]


def neural_network(input, weights):
    pred = ele_mul(input, weights)
    return pred


wlrec = [0.65, 1.0, 1.0, 0.9]

hurt = [0.1, 0.0, 0.0, 0.1]
win = [1, 1, 0, 1]
sad = [0.1, 0.0, 0.1, 0.2]

input = wlrec[0]

true = [hurt[0], win[0], sad[0]]

pred = neural_network(input, weights)

error = [0, 0, 0]
delta = [0, 0, 0]

for i in range(len(true)):
    error[i] = (pred[i] - true[i]) ** 2
    delta[i] = pred[i] - true[i]

print("Prediction:", pred)
print("Error:", error)
print("Delta:", delta)
```

## 8. Multiple Inputs dan Multiple Outputs

Pada tahap ini, neural network memiliki:

- beberapa input
- beberapa output
- banyak weight

Gradient descent tetap menggunakan prinsip yang sama.

Setiap weight memiliki weight_delta yang menunjukkan bagaimana weight
tersebut harus diubah untuk mengurangi error.

Dengan cara ini, gradient descent dapat digunakan pada network yang
lebih besar.

```python
def vect_mat_mul(vect, matrix):
    assert(len(vect) == len(matrix))

    output = [0, 0, 0]

    for i in range(len(vect)):
        output[i] = w_sum(vect, matrix[i])

    return output


def neural_network(input, weights):
    pred = vect_mat_mul(input, weights)
    return pred


weights = [
    [0.1, 0.1, -0.3],
    [0.1, 0.2, 0.0],
    [0.0, 1.3, 0.1]
]

input = [toes[0], wlrec[0], nfans[0]]

pred = neural_network(input, weights)

print("Prediction:", pred)
```

## 9. Apa yang Dipelajari oleh Weight?

Setiap weight berusaha mengurangi error.

Dalam dataset yang lebih besar, weight dapat merepresentasikan hubungan
antara input dan output.

Buku kemudian memperkenalkan dataset MNIST sebagai contoh nyata.

MNIST terdiri dari gambar angka tulisan tangan berukuran 28 × 28 piksel.
Artinya, setiap gambar dapat direpresentasikan sebagai 784 nilai input.

Network kemudian memiliki 10 output yang mewakili angka 0 sampai 9.

Dengan demikian:

784 input → 10 output

Weight belajar menemukan hubungan antara nilai pixel dan angka yang
diprediksi.

## 10. Visualisasi Weight dan Dot Product

Weight dapat divisualisasikan untuk melihat hubungan antara input dan output.

Pada klasifikasi gambar, setiap output mempunyai weight untuk setiap pixel.

Weight yang tinggi menunjukkan hubungan yang lebih kuat antara pixel
tersebut dengan output tertentu.

Selain itu, dot product dapat digunakan untuk melihat kemiripan antara
dua vector.

Secara sederhana:

dot product = jumlah dari perkalian setiap elemen dua vector.

Semakin mirip dua vector, semakin besar nilai dot product-nya.

```python
a = [0, 1, 0, 1]
b = [1, 0, 1, 0]

score = 0

for i in range(len(a)):
    score += a[i] * b[i]

print("Dot product:", score)
```
