# Chapter 15 — Deep Learning on Unseen Data

Pada chapter ini kita mempelajari **Federated Learning**, yaitu metode
melatih model machine learning tanpa harus mengumpulkan seluruh data
pengguna ke satu server.

Tujuan:
- Memahami masalah privasi pada deep learning.
- Memahami konsep federated learning.
- Memahami cara training model secara terdistribusi.
- Memahami secure aggregation.
- Mengenal homomorphic encryption.

## 1. Masalah Privasi pada Deep Learning

Dalam machine learning biasa:

User Data
   ↓
Server
   ↓
Training Model

Masalahnya adalah data pribadi harus dikirim dan dikumpulkan
di server.

Contohnya:
- Email
- Pesan
- Foto
- Data kesehatan
- Data aktivitas pengguna

Semakin banyak data dikumpulkan, semakin besar risiko privasi.

## 2. Federated Learning

**Federated Learning** adalah pendekatan machine learning di mana **data tetap berada di perangkat pengguna**, sedangkan model dikirim ke perangkat tersebut untuk dilatih.

### Alur Federated Learning

```text
                    SERVER
                       │
                  Model Awal
                       │
          ┌────────────┼────────────┐
          ↓            ↓            ↓
       USER 1        USER 2       USER 3
          │            │            │
      Training     Training     Training
          │            │            │
       Update 1     Update 2     Update 3
          │            │            │
          └────────────┼────────────┘
                       ↓
                    SERVER
                       │
                Gabungkan Update
                       │
                       ↓
                   Model Baru

## 3. Contoh Federated Learning

Misalnya kita ingin membuat model untuk mendeteksi spam email.

User 1 memiliki:
100 email

User 2 memiliki:
200 email

User 3 memiliki:
150 email

Pada machine learning biasa:
→ Semua email dikirim ke server.

Pada federated learning:
→ Model dikirim ke setiap user.
→ Model dilatih menggunakan email lokal.
→ User mengirim hasil/update model.
→ Server menggabungkan update tersebut.
→ Data email tetap berada di perangkat.

````python
import numpy as np

# Model awal
model = np.array([1.0, 2.0, 3.0])

# Hasil training dari beberapa user
user1 = np.array([1.2, 2.1, 3.1])
user2 = np.array([0.9, 2.3, 2.8])
user3 = np.array([1.1, 1.9, 3.2])

# Server menggabungkan update
new_model = (user1 + user2 + user3) / 3

print("Model awal :", model)
print("Model baru :", new_model)
````

## 4. Kelebihan dan Kekurangan

### Kelebihan
- Data pribadi tetap berada di perangkat.
- Mengurangi kebutuhan mengumpulkan data mentah.
- Dapat menggunakan data dari banyak perangkat.

### Kekurangan
- Perangkat memiliki kemampuan komputasi berbeda.
- Koneksi jaringan dapat menjadi masalah.
- Update model masih dapat memiliki risiko keamanan.
- Server perlu menggabungkan banyak update.

## 5. Secure Aggregation

Masalah berikutnya:

Walaupun data asli tidak dikirim,
update model dari setiap user masih dapat memberikan informasi.

Secure aggregation digunakan agar server hanya mengetahui
**hasil gabungan update**, bukan update masing-masing user.

Contoh:

User 1 → Update A
User 2 → Update B
User 3 → Update C

Server hanya mendapatkan:

(A + B + C)

bukan A, B, dan C secara terpisah.

## 6. Homomorphic Encryption

Homomorphic Encryption memungkinkan operasi tertentu dilakukan
pada data yang masih terenkripsi.

Secara sederhana:

Data asli
   ↓
Encrypt
   ↓
Data terenkripsi
   ↓
Operasi
   ↓
Hasil terenkripsi
   ↓
Decrypt
   ↓
Hasil akhir

Jadi pihak yang melakukan operasi tidak perlu melihat
data aslinya.

## 7. Kesimpulan

Machine Learning biasa:
Data → Server → Training

Federated Learning:
Model → User → Training lokal → Update → Server

Konsep penting:

1. Federated Learning
   → Training dilakukan pada perangkat pengguna.

2. Secure Aggregation
   → Server hanya melihat hasil gabungan update.

3. Homomorphic Encryption
   → Data dapat diproses dalam bentuk terenkripsi.
