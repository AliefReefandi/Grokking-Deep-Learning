## 3. Mengukur Error

Setelah menghasilkan prediction, kita perlu mengetahui apakah prediction
tersebut sudah benar atau masih memiliki kesalahan.

Salah satu cara sederhana untuk menghitung error adalah:

error = (prediction - actual)²

Metode ini disebut **mean squared error (MSE)** ketika digunakan untuk
menghitung rata-rata squared error dari beberapa data.

Pada contoh sederhana dalam chapter ini, error digunakan untuk melihat
seberapa jauh prediction dari target yang diinginkan.

Kuadrat digunakan agar error selalu bernilai positif dan kesalahan
positif maupun negatif tidak saling menghilangkan.

```python
weight = 0.1

def neural_network(input, weight):
    prediction = input * weight
    return prediction

number_of_toes = [8.5]
win_or_lose_binary = [1]

input = number_of_toes[0]
true = win_or_lose_binary[0]

pred = neural_network(input, weight)

error = (pred - true) ** 2

print("Prediction:", pred)
print("Error:", error)
```

### Penjelasan

Input = 8.5
Weight = 0.1

Prediction:

8.5 × 0.1 = 0.85

Target sebenarnya adalah 1.

Maka:

Error = (0.85 - 1)²
      = 0.0225

Semakin kecil nilai error, semakin dekat prediction dengan target.

## 5. Hot-and-Cold Learning

Salah satu metode sederhana untuk membuat neural network belajar adalah
**hot-and-cold learning**.

Ide dasarnya adalah mencoba mengubah weight ke dua arah:

- Weight dinaikkan.
- Weight diturunkan.

Kemudian error dari kedua kondisi tersebut dibandingkan.

Jika menaikkan weight menghasilkan error yang lebih kecil, weight dinaikkan.

Jika menurunkan weight menghasilkan error yang lebih kecil, weight diturunkan.

Proses ini dilakukan berulang kali sampai error menjadi semakin kecil.

```python
weight = 0.1
lr = 0.01

input = 8.5
true = 1

pred = neural_network(input, weight)
error = (pred - true) ** 2

# Mencoba weight yang lebih besar
p_up = neural_network(input, weight + lr)
e_up = (p_up - true) ** 2

# Mencoba weight yang lebih kecil
p_dn = neural_network(input, weight - lr)
e_dn = (p_dn - true) ** 2

print("Error awal :", error)
print("Error naik :", e_up)
print("Error turun:", e_dn)
```

### Penjelasan

Program mencoba dua kemungkinan:

1. `weight + lr`
2. `weight - lr`

Kemudian error dari kedua prediction dibandingkan dengan error awal.

Cara ini dapat menentukan arah perubahan weight, tetapi masih memiliki
keterbatasan karena harus mencoba beberapa kemungkinan secara manual.

Gradient descent dikembangkan untuk menentukan arah dan besar perubahan
weight dengan lebih efisien.

## 6. Gradient Descent

Gradient descent merupakan metode untuk mencari nilai weight yang dapat
mengurangi error.

Daripada mencoba weight ke atas dan ke bawah secara manual, gradient
descent menggunakan **derivative** untuk mengetahui arah perubahan weight.

Prinsipnya:

- Jika gradient bernilai positif, weight dikurangi.
- Jika gradient bernilai negatif, weight ditambah.

Secara umum:

weight = weight - derivative

Tujuannya adalah bergerak menuju titik dengan error paling kecil.

## 7. Derivative

Derivative digunakan untuk mengetahui hubungan antara perubahan weight
dan perubahan error.

Sederhananya, derivative menunjukkan:

"Jika weight diubah, seberapa besar dan ke arah mana error akan berubah?"

Jika derivative:

- Positif → weight perlu dikurangi.
- Negatif → weight perlu ditambah.
- Mendekati 0 → perubahan weight sudah semakin kecil.

Dengan informasi tersebut, neural network dapat memperbarui weight
secara lebih terarah.

## 8. Implementasi Gradient Descent

Pada contoh berikut, weight diperbarui menggunakan gradient descent.

Proses yang dilakukan pada setiap iterasi:

1. Membuat prediction.
2. Menghitung error.
3. Menghitung delta.
4. Menghitung perubahan weight.
5. Memperbarui weight.
6. Mengulangi proses tersebut.

```python
weight = 0.0

goal_pred = 0.8
input = 1.1

for iteration in range(4):
    pred = input * weight

    error = (pred - goal_pred) ** 2

    delta = pred - goal_pred

    weight_delta = delta * input

    weight = weight - weight_delta

    print(
        "Error:", error,
        "Prediction:", pred,
        "Weight:", weight
    )
```

### Penjelasan

Variabel `delta` menunjukkan seberapa jauh prediction dari target.

Kemudian:

weight_delta = delta × input

Nilai `weight_delta` digunakan untuk menentukan perubahan weight.

Weight kemudian diperbarui menggunakan:

weight = weight - weight_delta

Dengan melakukan proses tersebut berulang kali, prediction bergerak
mendekati target dan error semakin kecil.

## 9. Learning = Mengurangi Error

Tujuan utama proses learning adalah mengurangi error.

Pada setiap iterasi:

Prediction
    ->
Hitung Error
    ->
Hitung Gradient
    ->
Update Weight
    ->
Prediction Baru
    ->
Error Lebih Kecil

Jika proses learning berjalan dengan baik, nilai error akan semakin
mendekati 0.

Dengan demikian, neural network secara bertahap menemukan weight yang
menghasilkan prediction yang lebih dekat dengan target.

## 10. Overshooting dan Divergence

Gradient descent tidak selalu berjalan dengan baik.

Jika perubahan weight terlalu besar, network dapat melewati titik
minimum error.

Fenomena ini disebut **overshooting**.

Jika terjadi berulang kali, error justru semakin besar dan network
semakin menjauh dari target. Kondisi ini disebut **divergence**.

Salah satu penyebabnya adalah input yang terlalu besar sehingga
perubahan weight menghasilkan perubahan prediction yang sangat besar.

## 11. Alpha / Learning Rate

Untuk mengurangi overshooting, digunakan parameter yang disebut **alpha**.

Alpha digunakan untuk mengontrol seberapa besar perubahan weight pada
setiap iterasi.

Rumusnya:

weight = weight - (alpha × derivative)

Jika alpha terlalu besar:
- Perubahan weight terlalu besar.
- Network dapat melakukan overshooting.
- Error dapat meningkat atau mengalami divergence.

Jika alpha terlalu kecil:
- Perubahan weight menjadi sangat kecil.
- Learning berjalan lebih lambat.

Oleh karena itu, alpha perlu dipilih dengan nilai yang sesuai.

```python
weight = 0.5

goal_pred = 0.8
input = 2

alpha = 0.1

for iteration in range(20):
    pred = input * weight

    error = (pred - goal_pred) ** 2

    derivative = input * (pred - goal_pred)

    weight = weight - (alpha * derivative)

    print(
        "Error:", error,
        "Prediction:", pred
    )
```

### Penjelasan

Pada kode tersebut, `alpha = 0.1` digunakan untuk memperkecil perubahan
weight pada setiap iterasi.

Tanpa alpha, perubahan weight dapat terlalu besar sehingga menyebabkan
overshooting.

Dengan alpha, proses pembelajaran menjadi lebih terkontrol sehingga
prediction secara bertahap mendekati target.

Seiring bertambahnya iterasi, error semakin kecil dan prediction
mendekati nilai target yaitu 0.8.
